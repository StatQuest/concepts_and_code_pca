# Iris PCA (R)

-----

## Setup

```r
# load-packages
# Load ggplot2 for plotting
library(ggplot2)
```

```r
# load-data and print first few rows
head(iris)
```

```r
# Keep only the four numeric measurement columns (drop Species)
iris_num <- iris[, 1:4]
# print the first few rows of the numeric values
head(iris_num)
```

## Run PCA

```r
# run-pca
# Run PCA, centering and scaling each variable to unit variance
pca <- prcomp(iris_num, center = TRUE, scale. = TRUE)

# Print standard deviations and rotation (loadings) of each component
pca
```

```r
# pca-summary
# Show proportion of variance explained by each component
summary(pca)
# NOTE: summary() computes the proportion of variance 
# for each PC, but doesn't store it. So, later on, we'll
# have to calculate it ourselves.
```

## Variance explained

```r
# compute-variance
# Compute variance explained (squared sdev divided by total variance)
var_explained <- pca$sdev^2 / sum(pca$sdev^2)
# print out
var_explained
```

```r
# Build a data frame for the scree plot
scree_df <- data.frame(PC = paste0("PC", seq_along(var_explained)), Variance = var_explained)
```

```r
# scree-plot
# Plot the scree plot as bars of variance explained per component
ggplot(scree_df, aes(x = PC, y = Variance)) +
  # Draw one bar per principal component
  geom_col() +
  # Label the y-axis
  labs(y = "Proportion of variance explained")
```

## Sample scores

```r
# pc-scores-plot
# Combine PC scores (first two components) with the species labels
scores <- data.frame(pca$x[, 1:2], Species = iris$Species)

# Plot samples in PC1 vs PC2 space, colored by species
ggplot(scores, aes(x = PC1, y = PC2, color = Species)) +
  # Draw one point per flower
  geom_point() +
  # Label axes with percent variance explained
  labs(x = sprintf("PC1 (%.1f%%)", 100 * var_explained[1]),
       y = sprintf("PC2 (%.1f%%)", 100 * var_explained[2]))
```

```r
# Inspect loadings to see which features drive PC1 and PC2
round(pca$rotation[, 1:2], 3)
```
